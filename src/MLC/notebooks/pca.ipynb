# Principal Component Analysis (PCA) from Scratch
This notebook demonstrates how to implement Principal Component Analysis (PCA) from scratch using NumPy. We will build a simple PCA class, fit it to random data, and reduce the dimensionality of the dataset.

## 1. Import Required Libraries
We start by importing NumPy, which will be used for all matrix operations.

```python
import numpy as np
```

## 2. Implement the PCA Class
Below is a custom implementation of the PCA algorithm. The class includes methods to fit the model to the data, transform the data to a lower-dimensional space, and combine both steps. Variable names and comments have been improved for clarity.

```python
class PrincipalComponentAnalysis:
    """
    Principal Component Analysis (PCA) implementation from scratch.
    """
    def __init__(self, n_components: int):
        """
        n_components: Number of principal components to keep.
        """
        self.n_components = n_components
        self.projection_matrix = None
        self.mean_vector = None
    
    def fit(self, X: np.ndarray):
        """
        Fit the PCA model to the data X.
        """
        # Compute the mean of each feature
        self.mean_vector = np.mean(X, axis=0)
        # Center the data
        X_centered = X - self.mean_vector
        # Compute the covariance matrix
        covariance_matrix = X_centered.T @ X_centered / (X_centered.shape[0] - 1)  # shape: (num_features, num_features)
        # Eigen decomposition of the covariance matrix
        eigenvalues, eigenvectors = np.linalg.eigh(covariance_matrix)
        # Sort eigenvectors by descending eigenvalues
        sorted_indices = np.argsort(eigenvalues)[::-1]
        top_eigenvectors = eigenvectors[:, sorted_indices[:self.n_components]]
        self.projection_matrix = top_eigenvectors
        return self
    
    def transform(self, X: np.ndarray):
        """
        Project the data X onto the principal components.
        """
        X_centered = X - self.mean_vector
        return X_centered @ self.projection_matrix
    
    def fit_transform(self, X: np.ndarray):
        """
        Fit the model to X and return the transformed data.
        """
        self.fit(X)
        return self.transform(X)
```

## 3. Generate Sample Data
Let's create a random dataset with 100 samples and 5 features to demonstrate PCA.

```python
# Create a random dataset: 100 samples, 5 features
np.random.seed(42)  # For reproducibility
data = np.random.rand(100, 5)
```

## 4. Apply PCA and Reduce Dimensionality
Now, let's use our PCA implementation to reduce the dataset from 5 dimensions to 2. We will fit the model and transform the data, then print the shapes to verify the reduction.

```python
# Instantiate PCA to reduce to 2 dimensions
pca = PrincipalComponentAnalysis(n_components=2)
reduced_data = pca.fit_transform(data)
print('Original data shape:', data.shape)
print('Reduced data shape:', reduced_data.shape)
```
